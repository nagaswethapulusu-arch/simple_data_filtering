Simple Data Filtering

```python
# Student dataset represented as a list of dictionaries
students = [
    {"id": 101, "name": "Aarav",   "age": 20, "marks": 85, "department": "AI & ML"},
    {"id": 102, "name": "Bhavya",  "age": 22, "marks": 68, "department": "Data Science"},
    {"id": 103, "name": "Chirag",  "age": 19, "marks": 92, "department": "AI & ML"},
    {"id": 104, "name": "Divya",   "age": 21, "marks": 74, "department": "Computer Science"},
    {"id": 105, "name": "Eshwar",  "age": 23, "marks": 88, "department": "AI & ML"},
    {"id": 106, "name": "Fatima",  "age": 20, "marks": 59, "department": "Data Science"},
    {"id": 107, "name": "Gaurav",  "age": 22, "marks": 91, "department": "Computer Science"}
]

print(f"Total students loaded: {len(students)}")
```

```python
def print_records(title, records):
    """Utility function to print records in a neat table format."""
    print(f"\n{'='*20} {title} (Count: {len(records)}) {'='*20}")
    if not records:
        print("No records matched the condition.")
        return
    for record in records:
        print(f"ID: {record['id']} | Name: {record['name']:<8} | Age: {record['age']} | Marks: {record['marks']} | Dept: {record['department']}")

# Display all original students
print_records("All Student Records", students)
```

```python
# Filtering students who scored 80 or above
high_scorers = [s for s in students if s["marks"] >= 80]

print_records("Students with Marks >= 80", high_scorers)
```

```python
# Filtering students belonging to 'AI & ML' department
aiml_students = [s for s in students if s["department"] == "AI & ML"]

print_records("Students in AI & ML Department", aiml_students)
```

```python
# Filtering students aged 21 or younger
young_students = [s for s in students if s["age"] <= 21]

print_records("Students with Age <= 21", young_students)
```

```python
# Combining multiple conditions: AI & ML department AND marks >= 85 AND age <= 21
filtered_combined = [
    s for s in students 
    if s["department"] == "AI & ML" and s["marks"] >= 85 and s["age"] <= 21
]

print_records("AI & ML Students with Marks >= 85 & Age <= 21", filtered_combined)
```
